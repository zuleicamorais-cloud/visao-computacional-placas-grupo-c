### Etapa 02 — Localização dos arquivos do projeto

**Objetivo:** Identificar os arquivos e diretórios armazenados na pasta do projeto no Google Drive.

**Funcionamento:** `Path` define o caminho da pasta e `islice()` limita a listagem aos 10 primeiros itens.

**Validação:** Identificados `annotations.xml` e `images`.



```python
from pathlib import Path
from itertools import islice

pasta = Path('/content/drive/MyDrive/DADOS/Deteccao_Placas')

for arquivo in islice(pasta.iterdir(), 10):
    print(arquivo.name)
```

```python
pasta_imagens = pasta / 'images'

for arquivo in islice(pasta_imagens.iterdir(), 5):
    print(arquivo.name)
```

```python
from PIL import Image
import matplotlib.pyplot as plt

arquivo_imagem = next(p for p in pasta_imagens.iterdir() if p.suffix.lower() in ('.jpg', '.jpeg', '.png'))

img = Image.open(arquivo_imagem).convert('RGB')

print('Arquivo:', arquivo_imagem.name)
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
import xml.etree.ElementTree as ET

raiz = ET.parse(pasta / 'annotations.xml').getroot()

anotacao = next((item for item in raiz.findall('image') if Path(item.get('name', '')).name == arquivo_imagem.name), None)

print(ET.tostring(anotacao, encoding='unicode') if anotacao is not None else 'Anotação não encontrada.')
```

```python
import numpy as np
from matplotlib.patches import Polygon

marca = anotacao.find('points')
pontos = np.array([list(map(float, par.split(','))) for par in marca.get('points').split(';')])

fig, ax = plt.subplots(figsize=(12, 7))
ax.imshow(img)
ax.add_patch(Polygon(pontos, closed=True, fill=False, edgecolor='lime', linewidth=2))
ax.set_title('Categoria anotada: ' + marca.get('label'))
ax.axis('off')
plt.show()
```

```python
esquerda, topo = np.floor(pontos.min(axis=0)).astype(int)
direita, base = np.ceil(pontos.max(axis=0)).astype(int)

placa = img.crop((esquerda, topo, direita, base))

print('Tamanho do recorte:', placa.size)
plt.imshow(placa)
plt.title('Categoria anotada: ' + marca.get('label'))
plt.axis('off')
plt.show()
```

```python
fig, telas = plt.subplots(1, 2, figsize=(12, 5))

telas[0].imshow(img)
telas[0].set_title('Fotografia original')
telas[0].axis('off')

telas[1].imshow(placa)
telas[1].set_title('Placa recortada — ' + marca.get('label'))
telas[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
placas = pd.DataFrame([{'arquivo': Path(r.get('name')).name, 'categoria': m.get('label'), 'pontos': m.get('points')} for r in registros_com_placas for m in r.findall('points')])

placas['quantidade_pontos'] = placas['pontos'].str.split(';').str.len()

print('Total de placas:', len(placas))
print('Quantidade de pontos por anotação:')
print(placas['quantidade_pontos'].value_counts().sort_index())

display(placas.head())
```

```python
caso = placas.loc[placas['quantidade_pontos'] == 1].iloc[0]

foto_caso = Image.open(arquivos_subpastas[caso['arquivo']]).convert('RGB')
pontos_caso = np.array([list(map(float, par.split(','))) for par in caso['pontos'].split(';')])

plt.figure(figsize=(12, 7))
plt.imshow(foto_caso)
plt.scatter(pontos_caso[:, 0], pontos_caso[:, 1], color='lime', s=80, marker='x')
plt.title('Anotação com um ponto — ' + caso['categoria'])
plt.axis('off')
plt.show()

print('Arquivo:', caso['arquivo'])
print('Coordenadas:', caso['pontos'])
```

```python
coordenadas_placas = [np.array([list(map(float, par.split(','))) for par in texto.split(';')]) for texto in placas['pontos']]

placas['largura_anotada'] = [p[:, 0].max() - p[:, 0].min() for p in coordenadas_placas]
placas['altura_anotada'] = [p[:, 1].max() - p[:, 1].min() for p in coordenadas_placas]

placas['tem_area'] = (placas['largura_anotada'] > 0) & (placas['altura_anotada'] > 0)

print('Anotações com largura e altura positivas:', placas['tem_area'].sum())
print('Anotações sem largura ou sem altura:', (~placas['tem_area']).sum())

display(placas.loc[~placas['tem_area'], ['arquivo', 'categoria', 'quantidade_pontos', 'largura_anotada', 'altura_anotada']].head())
```

```python
dimensoes_xml = {Path(r.get('name')).name: (int(r.get('width')), int(r.get('height'))) for r in registros_com_placas}

placas['dentro_limites_xml'] = [bool(np.isfinite(p).all() and (p >= 0).all() and (p[:, 0] <= dimensoes_xml[nome][0]).all() and (p[:, 1] <= dimensoes_xml[nome][1]).all()) for nome, p in zip(placas['arquivo'], coordenadas_placas)]

print('Dentro dos limites do XML:', placas['dentro_limites_xml'].sum())
print('Precisam de revisão:', (~placas['dentro_limites_xml']).sum())
```

```python
from tqdm.auto import tqdm
```

```python
from tqdm.auto import tqdm

conferencia_imagens = []

for nome, tamanho_xml in tqdm(dimensoes_xml.items(), desc='Conferindo fotografias'):
    try:
        with Image.open(arquivos_subpastas[nome]) as foto:
            tamanho_real = foto.size

        status = 'confere' if tamanho_real == tamanho_xml else 'dimensao_diferente'
        conferencia_imagens.append({'arquivo': nome, 'status': status, 'detalhe': str(tamanho_real)})

    except Exception as erro:
        conferencia_imagens.append({'arquivo': nome, 'status': 'erro_abertura', 'detalhe': str(erro)})

auditoria_imagens = pd.DataFrame(conferencia_imagens)

print(auditoria_imagens['status'].value_counts())

auditoria_imagens.to_csv(pasta_resultados / 'auditoria_dimensoes.csv', index=False, encoding='utf-8-sig')
```

```python
arquivos_conferidos = set(auditoria_imagens.loc[auditoria_imagens['status'] == 'confere', 'arquivo'])

criterio_recorte = placas['tem_area'] & placas['dentro_limites_xml'] & placas['arquivo'].isin(arquivos_conferidos)

base_recortes = placas.loc[criterio_recorte].copy()
pendencias = placas.loc[~criterio_recorte].copy()

base_recortes.to_csv(pasta_resultados / 'base_para_recortes.csv', index=True, index_label='id_anotacao', encoding='utf-8-sig')
pendencias.to_csv(pasta_resultados / 'anotacoes_pendentes.csv', index=True, index_label='id_anotacao', encoding='utf-8-sig')

print('Anotações para tentar recortar:', len(base_recortes))
print('Anotações pendentes:', len(pendencias))
```

```python
amostra_revisao = base_recortes.groupby('categoria').sample(n=1, random_state=42)

fig, telas = plt.subplots(len(amostra_revisao), 2, figsize=(12, 4 * len(amostra_revisao)), squeeze=False)

for linha, (id_anotacao, dados) in enumerate(amostra_revisao.iterrows()):
    with Image.open(arquivos_subpastas[dados['arquivo']]) as original:
        foto = original.convert('RGB')

    pontos = np.array([list(map(float, par.split(','))) for par in dados['pontos'].split(';')])
    limites = tuple(np.floor(pontos.min(axis=0)).astype(int)) + tuple(np.ceil(pontos.max(axis=0)).astype(int))
    recorte = foto.crop(limites)

    telas[linha, 0].imshow(foto)
    telas[linha, 0].add_patch(Polygon(pontos, closed=True, fill=False, edgecolor='lime', linewidth=2))
    telas[linha, 0].set_title(f'Anotação {id_anotacao}')

    telas[linha, 1].imshow(recorte)
    telas[linha, 1].set_title(f'{dados["categoria"]} — {recorte.width} × {recorte.height} pixels')

    telas[linha, 0].axis('off')
    telas[linha, 1].axis('off')

plt.tight_layout()
plt.show()
```

```python
fig.savefig(pasta_resultados / 'comparacao_por_categoria.png', dpi=150, bbox_inches='tight')

amostra_revisao.to_csv(
    pasta_resultados / 'amostra_revisao_por_categoria.csv',
    index=True,
    index_label='id_anotacao',
    encoding='utf-8-sig'
)

print('Comparação e lista de exemplos salvas.')
```

```python
lote_piloto = base_recortes.sample(frac=1, random_state=42).groupby('categoria').head(10)

pasta_piloto = pasta_resultados / 'recortes_piloto'
pasta_piloto.mkdir(parents=True, exist_ok=True)

resultado_piloto = []

for identificador, dados in tqdm(lote_piloto.iterrows(), total=len(lote_piloto), desc='Salvando recortes'):
    try:
        pontos = np.array([list(map(float, par.split(','))) for par in dados['pontos'].split(';')])
        limites = tuple(np.floor(pontos.min(axis=0)).astype(int)) + tuple(np.ceil(pontos.max(axis=0)).astype(int))
        destino = pasta_piloto / f'placa_{identificador:05d}.png'

        with Image.open(arquivos_subpastas[dados['arquivo']]) as foto:
            recorte = foto.convert('RGB').crop(limites)
            recorte.save(destino)

        resultado_piloto.append({'id_anotacao': identificador, 'arquivo_origem': dados['arquivo'], 'categoria': dados['categoria'], 'status': 'salvo', 'detalhe': destino.name})

    except Exception as erro:
        resultado_piloto.append({'id_anotacao': identificador, 'arquivo_origem': dados['arquivo'], 'categoria': dados['categoria'], 'status': 'erro', 'detalhe': str(erro)})

relatorio_piloto = pd.DataFrame(resultado_piloto)
relatorio_piloto.to_csv(pasta_resultados / 'relatorio_recortes_piloto.csv', index=False, encoding='utf-8-sig')

print(relatorio_piloto['status'].value_counts())
print('Pasta:', pasta_piloto)
```

```python
salvos_piloto = relatorio_piloto.loc[relatorio_piloto['status'] == 'salvo']

quantidade_linhas = (len(salvos_piloto) + 5) // 6
fig_piloto, telas_piloto = plt.subplots(quantidade_linhas, 6, figsize=(18, 3 * quantidade_linhas))
telas_piloto = np.asarray(telas_piloto).ravel()

for tela in telas_piloto:
    tela.axis('off')

for tela, (_, dados) in zip(telas_piloto, salvos_piloto.iterrows()):
    with Image.open(pasta_piloto / dados['detalhe']) as imagem:
        tela.imshow(imagem.copy())
    tela.set_title(f'ID {dados["id_anotacao"]}\n{dados["categoria"]}', fontsize=9)

plt.tight_layout()
plt.show()
```
